```python
from transformers import pipeline
```

```python
generator = pipeline("text-generation", model="gpt2")
import textwrap
prompt = "Artificial Intelligence will change the future because"

result = generator(prompt, max_new_tokens=50,truncation=True)

print("Generated Text:\n")
print(textwrap.fill(result[0]['generated_text']))
```

```python
generator1 = pipeline("text-generation", model="distilgpt2")
prompt = "Climate change is a global challenge because"

result = generator1(prompt, max_new_tokens=50)
print("Generated Text:\n")
print(textwrap.fill(result[0]['generated_text'],width=100))
```

```python
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

text = """
Artificial Intelligence is one of the fastest growing technologies in the world.
It is widely used in healthcare, finance, transportation and education.
AI systems can analyze huge amounts of data and help humans make better decisions.
However, there are also concerns about privacy, job loss and ethical issues.
Researchers and governments are working to ensure that AI is used responsibly.
"""

summary = summarizer(text, max_length=50, min_length=20)

print("Summary:\n")
print(textwrap.fill(summary[0]['summary_text'],width=100))
```

```python
text = """
Machine learning is a branch of artificial intelligence that allows computers
to learn from data without being explicitly programmed. It is used in many
applications such as recommendation systems, fraud detection, speech
recognition, and image classification. With the growth of big data and
computing power, machine learning has become a very important technology.
"""

summary = summarizer(text, max_length=40, min_length=15)

print(summary[0]['summary_text'])
```

```python
summarizer2 = pipeline("text2text-generation", model="t5-small")

text = """
The Internet has changed the way people communicate and access information.
Today billions of people use the internet for education, entertainment,
online shopping and social networking. Businesses also depend heavily
on internet technologies to reach global customers. However, concerns
about data privacy and cybersecurity are increasing.
"""

summary = summarizer2("summarize: " + text, max_length=40)

print("Summary:")
print(textwrap.fill(summary[0]['generated_text'],width=100))
```

```python
qa = pipeline("question-answering", model="distilbert-base-cased-distilled-squad")

context = """
Hugging Face is a company that develops open source tools for natural language processing.
Its Transformers library provides thousands of pretrained models for tasks like text generation,
summarization, translation and question answering.
"""

question = "What library does Hugging Face provide?"

result = qa(question=question, context=context)

print("Question:", question)
print("Answer:", result['answer'])
```

```python
context = """
Google was founded in 1998 by Larry Page and Sergey Brin.
The company started as a search engine but later expanded into
many products like Gmail, Google Maps and Android.
"""

question = "Who founded Google?"

result = qa(question=question, context=context)

print("Answer:", result['answer'])
```

```python
qa2 = pipeline("question-answering", model="deepset/roberta-base-squad2")

context = """
Python is a popular programming language created by Guido van Rossum.
It was first released in 1991 and is widely used for web development,
data science, artificial intelligence and automation.
"""

question = "Who created Python?"

result = qa2(question=question, context=context)

print("Question:", question)
print("Answer:", result['answer'])
```
