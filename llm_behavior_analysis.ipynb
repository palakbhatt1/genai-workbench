````python
from groq import Groq
from google.colab import userdata

class LLMExperiment:
    def __init__(self):

        self.client = Groq(api_key=userdata.get("api_key"))

    def ask_model(self, model, system_prompt, user_prompt, temperature=0.5):
        response = self.client.chat.completions.create(
            model=model,
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": user_prompt}
            ],
            temperature=temperature,
            max_completion_tokens=512
        )
        return response.choices[0].message.content

    # ---------- Assignment 1 ----------
    def assignment_1(self):
        prompt = "Explain Artificial Intelligence to a first-year college student."

        models = {
            "GPT-OSS-120B": "openai/gpt-oss-120b",
            "GPT-OSS-SAFEGUARD-20B": "openai/gpt-oss-safeguard-20b",
            "LLaMA-3.1-8B-Instant": "llama-3.1-8b-instant"
        }

        outputs = {}
        for name, model in models.items():
            outputs[name] = self.ask_model(
                model,
                "You are a helpful college-level instructor.",
                prompt
            )

        return outputs

    # ---------- Assignment 2 ----------
    def assignment_2(self):
        base_question = "What is Machine Learning?"

        instruction_variants = {
            "2 lines": "Answer in 2 lines.",
            "Real-life example": "Explain using a real-life example.",
            "For a child": "Explain for a 10-year-old child."
        }

        model = "llama-3.1-8b-instant"
        outputs = {}

        for style, instruction in instruction_variants.items():
            prompt = f"{instruction} {base_question}"
            outputs[style] = self.ask_model(
                model,
                "You are a helpful teacher.",
                prompt
            )

        return outputs

    # ---------- Assignment 3 ----------
    def assignment_3(self):
        prompt = "List three applications of Artificial Intelligence in daily life."
        model = "llama-3.1-8b-instant"

        outputs = []
        for run in range(3):
            outputs.append(
                self.ask_model(
                    model,
                    "You are a helpful assistant.",
                    prompt,
                    temperature=0.8
                )
            )

        return outputs
````

````python
experiment = LLMExperiment()
````

````python
a1_outputs = experiment.assignment_1()
print("--- Assignment 1 Outputs ---")
for model_name, response_text in a1_outputs.items():
    print(f"\n### Model: {model_name}")
    print("```markdown")
    print(response_text)
    print("```")
    print("--------------------------------------------------")
````
